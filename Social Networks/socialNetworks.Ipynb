```python
%%manim Shapes


class Shapes(Scene):

    def construct(self):
        # Define los nodos del grafo
        nodes = {
            "A": [0, 0, 0],
            "B": [2, 1, 0],
            "C": [1, 3, 0],
            "D": [-2, 1, 0]
        }

        # Define las conexiones (aristas) entre los nodos
        edges = [
            ("A", "B"),
            ("B", "C"),
            ("C", "A"),
            ("A", "D")
        ]

        # Crea los nodos como círculos
        circles = {key: Circle(radius=0.3).move_to(value) for key, value in nodes.items()}

        # Muestra los círculos
        for circle in circles.values():
            self.play(Create(circle))

        # Añade etiquetas a los nodos
        for node, position in nodes.items():
            label = Text(node).move_to(position)
            self.play(Create(label))

        # Dibuja las aristas
        for start, end in edges:
            start_circle = circles[start]
            end_circle = circles[end]
            self.play(Create(self.get_line(start_circle.get_center(), end_circle.get_center())))

    def get_line(self, start, end):
        # Crea una línea entre dos puntos
        return Line(start, end)
```
